```python
"""Example script for importing variables into OpenSILEX."""
```

```python
import os

from opensilex_python_client.auth import connect
from opensilex_python_client.variables import import_from_csv
from opensilex_python_client.variables.groups import update
```

```python
# 1. Authentication
client = connect.connect_to_opensilex(
    {"host": "http://localhost:8666/rest", "identifier": "admin@opensilex.org", "password": "admin"}
)
```

```python
# 2. File paths (modify these paths)
save_file_path = os.path.join(os.path.abspath(""), "output_files")
os.makedirs(save_file_path, exist_ok=True)
input_csv_path = os.path.realpath(os.path.join(os.path.abspath(""),"../data_examples/variables", "test_variables.csv"))
input_config_path = os.path.realpath(os.path.join(os.path.abspath(""),"../data_examples/variables", "test_config.yaml"))
print(f"Input CSV path: {os.path.realpath(input_csv_path)}")
print(f"Input Config path: {os.path.realpath(input_config_path)}")
```

```python
# 3. Import variables (creates components + variables automatically)
grouped_vars = import_from_csv.run(client, input_csv_path, input_config_path)
```

```python
# 4. Attach to groups
update.attach_to_groups(client, grouped_vars, input_config_path)
```

```python
print(" Import completed!")
```
